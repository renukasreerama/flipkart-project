```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load datasets
orders_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\orders_dataset.csv")
order_items_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\order_items_dataset.csv")

# Merge orders and order_items
eda_df = pd.merge(orders_df, order_items_df, on='order_id', how='inner')

# Convert timestamp to datetime
eda_df['order_purchase_timestamp'] = pd.to_datetime(eda_df['order_purchase_timestamp'], errors='coerce')

# Create Year-Month for grouping
eda_df['Year-Month'] = eda_df['order_purchase_timestamp'].dt.to_period('M').astype(str)

# Group by Year-Month
monthly_eda = eda_df.groupby('Year-Month').agg(
    Total_Revenue=('price', 'sum'),
    Total_Orders=('order_id', 'nunique'),
    Average_Order_Value=('price', 'mean')
).reset_index()

# Calculate Month-on-Month % Change
monthly_eda['Revenue_MoM%'] = monthly_eda['Total_Revenue'].pct_change() * 100
monthly_eda['Orders_MoM%'] = monthly_eda['Total_Orders'].pct_change() * 100

# Plot trends with peak/dip annotations
sns.set(style="whitegrid")
plt.figure(figsize=(16, 8))

# Revenue line
sns.lineplot(data=monthly_eda, x='Year-Month', y='Total_Revenue', marker='o', label='Total Revenue', color='blue')
plt.axvline(x='2017-11', color='yellow', linestyle='--', alpha=0.5)
plt.axvline(x='2016-12', color='red', linestyle='--', alpha=0.5)
plt.text('2017-11', monthly_eda['Total_Revenue'].max()*1.05, 'Peak Revenue\nNov 2017', ha='center', color='blue')
plt.text('2016-12', monthly_eda['Total_Revenue'].max()*0.3, 'Lowest Revenue\nDec 2016', ha='center', color='red')

# Orders line
sns.lineplot(data=monthly_eda, x='Year-Month', y='Total_Orders', marker='s', label='Total Orders', color='green')
plt.axvline(x='2017-11', color='green', linestyle='--', alpha=0.5)
plt.axvline(x='2016-12', color='orange', linestyle='--', alpha=0.5)
plt.text('2017-11', monthly_eda['Total_Orders'].max()*1.05, 'Peak Orders\nNov 2017', ha='center', color='green')
plt.text('2016-12', monthly_eda['Total_Orders'].max()*0.3, 'Lowest Orders\nDec 2016', ha='center', color='orange')

# Final plot settings
plt.title("Monthly Revenue and Order Trends with Peaks and Dips", fontsize=16)
plt.xlabel("Year-Month")
plt.ylabel("Value")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt

# Load your data
orders_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\orders_dataset.csv")
order_items_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\order_items_dataset.csv")

# Merge datasets
merged_df = pd.merge(orders_df, order_items_df, on='order_id', how='inner')

# Convert to datetime
merged_df['order_purchase_timestamp'] = pd.to_datetime(merged_df['order_purchase_timestamp'], errors='coerce')

# Set index and resample revenue monthly
merged_df.set_index('order_purchase_timestamp', inplace=True)
monthly_revenue = merged_df['price'].resample('ME').sum().reset_index()
monthly_revenue.columns = ['ds', 'y']  # Prophet requires these column names

# Initialize and fit Prophet model
model = Prophet()
model.fit(monthly_revenue)

# Forecast next 24 months
future = model.make_future_dataframe(periods=24, freq='M')
forecast = model.predict(future)

# Plot forecast
model.plot(forecast)
plt.title("Monthly Revenue Forecast (Next 24 Months)")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid(True)
plt.tight_layout()
plt.show()

# Optional: Save forecast to CSV
# forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].to_csv("revenue_forecast.csv", index=False)
```

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt

# Load necessary datasets
orders_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\orders_dataset.csv")
order_items_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\order_items_dataset.csv")
products_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\products_dataset.csv")
product_category_name_translation_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\product_category_name_translation.csv")

# Merge necessary data
df = orders_df.merge(order_items_df, on="order_id", how="inner")
df = df.merge(products_df, on="product_id", how="inner")
df = df.merge(product_category_name_translation_df, on="product_category_name", how="left")

# Convert purchase timestamp
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])

# Optional: choose top N categories by total revenue
top_categories = (
    df.groupby("product_category_name_english")['price'].sum()
    .sort_values(ascending=False).head(3).index.tolist()
)

# Loop over each top category and forecast revenue
for category in top_categories:
    cat_df = df[df['product_category_name_english'] == category].copy()
    cat_df.set_index('order_purchase_timestamp', inplace=True)
    
    # Resample to monthly revenue
    monthly_cat_revenue = cat_df['price'].resample('M').sum().reset_index()
    monthly_cat_revenue.columns = ['ds', 'y']

    # Fit Prophet model
    model = Prophet()
    model.fit(monthly_cat_revenue)

    # Forecast next 12 months
    future = model.make_future_dataframe(periods=12, freq='ME')
    forecast = model.predict(future)

    # Plot forecast
    print(f"\n📦 Forecast for Category: {category}")
    model.plot(forecast)
    plt.title(f"Forecast - {category}", fontsize=14)
    plt.xlabel("Date")
    plt.ylabel("Revenue")
    plt.tight_layout()
    plt.grid(True)
    plt.show()
```

```python
import pandas as pd
from prophet import Prophet
import matplotlib.pyplot as plt

# Load datasets
orders_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\orders_dataset.csv")
order_items_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\order_items_dataset.csv")
products_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\products_dataset.csv")
category_translation_df = pd.read_csv(r"C:\Users\sreer\Downloads\unzipped_data\Mercado Livre - ecom\product_category_name_translation.csv")

# Merge data
df = orders_df.merge(order_items_df, on="order_id", how="inner")
df = df.merge(products_df, on="product_id", how="inner")
df = df.merge(category_translation_df, on="product_category_name", how="left")

# Convert date column
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'])

# Choose top 1 category for detailed analysis
top_category = (
    df.groupby("product_category_name_english")['price'].sum()
    .sort_values(ascending=False).index[0]
)

# Filter for the top category
cat_df = df[df['product_category_name_english'] == top_category].copy()
cat_df.set_index('order_purchase_timestamp', inplace=True)

# Create monthly revenue time series
monthly_cat_revenue = cat_df['price'].resample('ME').sum().reset_index()
monthly_cat_revenue.columns = ['ds', 'y']

# Fit Prophet model
model = Prophet()
model.fit(monthly_cat_revenue)

# Forecast next 12 months
future = model.make_future_dataframe(periods=12, freq='ME')
forecast = model.predict(future)

# Plot forecast
print(f"📦 Forecast for Category: {top_category}")
fig1 = model.plot(forecast)
plt.title(f"Forecast - {top_category}")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid(True)
plt.show()

# Plot components: trend, weekly & yearly seasonality
fig2 = model.plot_components(forecast)
plt.suptitle(f"Trend & Seasonality - {top_category}", fontsize=16)
plt.tight_layout()
plt.show()
```
